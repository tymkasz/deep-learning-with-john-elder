```python
import torch
```

```python
my_torch = torch.arange(10)
my_torch
```

```python
# Reshape and View
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
# Reshape if we don't know the number of items using -1
my_torch2 = torch.arange(18)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(3,-1)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(-1,6)
my_torch2
```

```python
# View
my_torch3 = torch.arange(10)
my_torch3
```

```python
my_torch4 = my_torch3.view(2,5)
my_torch4
```

[Differences between reshape() and view()](https://stackoverflow.com/questions/49643225/whats-the-difference-between-reshape-and-view-in-pytorch?rq=1)

```python
# With reshape and view, they will update
my_torch5 = torch.arange(10)
my_torch5
```

```python
my_torch6 = my_torch5.reshape(2,5)
my_torch6
```

```python
my_torch5[1] = 4232
my_torch5
```

```python
# Making changes to the original tensor, changes will be reflected
my_torch6
```

```python
# Slices
my_torch7 = torch.arange(10)
my_torch7
```

```python
# It is different than regular 7, it is kinda like scalar -> grab a specific item
my_torch7[7]
```

```python
# Grab slice
my_torch8 = my_torch7.reshape(5,2)
my_torch8
```

```python
my_torch8[:,1]
```

```python
# return column
my_torch8[:,1:]
```
