**TENSOR MATH OPERATIONS**
*   Add, substract, multiply, divide, remainders, exponents, modules
*   Shorthand and Longhand
*   Reassignment

```python
import torch
import numpy as np
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
```

```python
# Same amount of elements
# Add
tensor_a + tensor_b
```

```python
# Addition Longhand
torch.add(tensor_a, tensor_b)
```

```python
# Substract
tensor_a - tensor_b
```

```python
# Sub Function
torch.sub(tensor_a, tensor_b)
```

```python
# Multiply
tensor_a * tensor_b
```

```python
# Mul Function
torch.mul(tensor_a, tensor_b)
```

```python
# Divide
tensor_a / tensor_b
```

```python
# Div Longhand
torch.div(tensor_a, tensor_b)
```

```python
# Remainder Modulus
tensor_b % tensor_a
## Divided by corresponding element
```

```python
# Remainder Longhand
torch.remainder(tensor_b, tensor_a)
```

```python
# Exponent (power)
tensor_a ** tensor_b
## Power of coressponding element
```

```python
# Power longhand
torch.pow(tensor_a, tensor_b)
```

```python
# Another way to write longhand
tensor_a.add(tensor_b)
tensor_a.sub(tensor_b)
tensor_a.mul(tensor_b)
tensor_a.div(tensor_b)
tensor_a.remainder(tensor_b)
tensor_a.pow(tensor_b)
```

```python
# Reassignment
tensor_a + tensor_b
```

```python
# tensor_a = tensor_a + tensor_b - the same as below -> _
tensor_a.add_(tensor_b)
tensor_a
```
